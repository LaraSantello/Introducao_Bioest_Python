### Modelos

Em ciência, estatística, engenharia e IA, **modelos** são **representações simplificadas da realidade**, usadas para **entender, explicar, prever ou controlar** um fenômeno.

---

## 📌 Definição essencial

> **Um modelo é uma abstração da realidade**  
> **Espera-se que este preserve os aspectos mais relevantes da idealização pesquisada.**

Ou, de forma ainda mais direta:

> **Modelo = realidade simplificada + propósito**

## 🧱 Estatística

### 1️⃣ Limitações

Por não podermos acessar `todos os dados`, temos que coletar amostras e calcular valores médios (via modelos) amostrais e se possível inferir os valores médios populacionais. Sempre que medimos algo implicitamente estamos comentendo erros. Somente podemos calcular erros ou desvios se medirmos várias vezes. Logo, medidas amostrais resultam em um valor esperado (media) e uma medidade de erro-dispersão (desvio padrão). Esta é a base da estatística clássica, além disto ela presume que só teremos bons resultados se aumentarmos muito o número amostral, ou seja, a quantidade de amostras. Fato, que muitas vezes se apresenta como uma limitação. Para tanto, a estatística clássica desenvolveu métodos que resultam em erros maiores para números amostrais menores, e vice-versa.

Caso os valores médios calculados (esperanças) se aproximem dos observados (hipotética realidade, vide Kant), diz-se que o modelo é bom ou aceitável

## 📌 Nunca temos certeza de nada

Nunca temos certeza de nada, logo não há medição exata. Caso você apresente um resultado somente com o valor esperado, estará incorrendo em erro! Ou seja, um resultado científico (média ou valor esperado) tem que vir seguido de sua respectiva incerteza (desvio padrão). Se quisermos quantificar a incerteza, temos que utilizar o intervalo de confiança. Todos estes conceitos serão abordados neste curso.

### 3️⃣ Modelos teóricos

Por vezes não podemos acessar os dados para fazer amostragem, ou não os entedemos, nestes casos só nos resta criar um modelo e fazer simulações. Após a simulação, de posse dos dados simulados, analisamos se estes se aproximam de macrovalores observados. Lembre-se do modelo dos gases perfeitos modelados por Boltzmann e Maxwell; lembre-se da fórmula de gases ideais: PV = nRT.

Ou seja, caso obtenhamos os dados simulados, via um modelo proposto, poderemos inferir seus parâmetros e possível comportamento através da média e do desvio padrão. Simulando as moléculas de um gás ideal, Boltzmann e Maxwell propuseram uma estatística para a velocidade das partículas (gás ideal). Esta estatística permitia inferir a temperatura do gás. Logo, através de um modelo, pode-se associar Velocidade a Temperatura, ou seja, a Temperatura é um fenômeno macro advindo da movimentação molecular. Este modelo foi sugerido e a teoria proposta numa época em que não se sabia sequer se átomos e moléculas existiam.

### 4️⃣ Algo pode dar errado: não esconda

Nunca oculte nada em ciência! Ciência tem que ser reprodutível, logo, nunca omita a quantidade de amostras coletadas (n), o momento em que foi coletada (data/hora), equipamento, dados complementares (pessoas, local, temperatura, humidade, etc) e métodos.

Provavelmente, se o experimento fracassou, você não o publicará. Há raros casos de falhas que geraram artigos de interesse da comunidade científica (p.ex., Michelson & Morley e a medição do éter que resultou em fracasso).

Já se seu experimento deu certo, todos os dados, modelos, equipamentos, insumos e códigos-fonte computacionais têm que ser apresentados, para possível confirmação e reprodução.

---

### 5️⃣ Viés é um tipo de erro; falsidade é má conduta

Caso se manipulem dados, não adianta fazer estatística sobre os mesmos. Ao se colocar dados ruins numa máquina de inferência, extrair-se-ão automaticamente dados ruins — e ninguém conseguirá reproduzir seu experimento.

Por vezes, não calibramos corretamente um equipamento e ele gera uma leitura com o valor esperado + um viés. Este é um erro que não pode acontecer. Se gerar resultados significativos, você irá escrever um artigo, impossível de ser reproduzido. Se o artigo passar por um revisor e for editado, algum outro pesquisador alertará sobre o erro e o artigo terá que ser excluído e uma carta de retratação será necessária para a continuação dos trabalhos da equipe que produziu estes resultados.

Já, falsificar dados é crime e má conduta.

## 🎯 O critério correto para julgar um modelo

Não é:

* “é realista?”
* “é verdadeiro?”

Mas sim:

* **Para quê ele serve?**
* **Em que condições funciona?**
* **O que ele ignora?**
